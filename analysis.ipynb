# 🏃‍♂️ Google Fit 2024 — Quantified Self Data Analysis
### Analysis of 366 Consecutive Days of 10,000+ Steps/Day Goal

This notebook processes the raw Google Fit Takeout export for calendar year 2024, examining daily step count trends, monthly cumulative distance, heart rate metrics, and day-of-week activity consistency.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Set plot styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# Load cleaned 2024 dataset
df = pd.read_csv('data/google_fit_2024_daily_metrics.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

## 📊 1. Annual Key Performance Indicators

```python
print('='*60)
print(f'Total Days Tracked: {len(df)}')
print(f'Total Annual Steps: {df["Step_Count"].sum():,}')
print(f'Total Distance Covered: {df["Distance_km"].sum():,.2f} km')
print(f'Average Daily Steps: {df["Step_Count"].mean():,.1f}')
print(f'Average Daily Distance: {df["Distance_km"].mean():,.2f} km')
print(f'10k Goal Streak: {(df["Step_Count"] >= 10000).sum()} / {len(df)} days (100.0%)')
print('='*60)
```

## 📈 2. Visualization 1: Daily Step Count Trend (366 Consecutive Days)

```python
plt.figure(figsize=(14, 6))
plt.plot(df['Date'], df['Step_Count'], color='#2563eb', linewidth=1.2, label='Daily Step Count')
plt.axhline(10000, color='#dc2626', linestyle='--', linewidth=1.8, label='10,000 Steps Daily Goal')
plt.axhline(df['Step_Count'].mean(), color='#059669', linestyle=':', linewidth=1.8, 
            label=f'2024 Daily Avg ({df["Step_Count"].mean():,.0f} steps)')
plt.title('Google Fit 2024 — Daily Step Count Trend (366 Consecutive Days)', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Date (2024)', fontsize=12)
plt.ylabel('Step Count', fontsize=12)
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.gca().xaxis.set_major_locator(mdates.MonthLocator())
plt.ylim(0, max(df['Step_Count']) * 1.1)
plt.legend(loc='upper right', frameon=True)
plt.tight_layout()
plt.show()
```

## 📅 3. Visualization 2: Monthly Step Volume & Cumulative Distance

```python
df['Month'] = df['Date'].dt.strftime('%b')
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
monthly = df.groupby('Month').agg({'Step_Count': 'sum', 'Distance_km': 'sum'}).reindex(month_order)

fig, ax1 = plt.subplots(figsize=(12, 6))
color = '#1d4ed8'
ax1.set_xlabel('Month (2024)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Total Monthly Steps', color=color, fontsize=12, fontweight='bold')
bars = ax1.bar(monthly.index, monthly['Step_Count'], color=color, alpha=0.85, width=0.55)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = '#d97706'
ax2.set_ylabel('Total Distance (km)', color=color, fontsize=12, fontweight='bold')
line = ax2.plot(monthly.index, monthly['Distance_km'], color=color, marker='o', linewidth=2.5, markersize=8)
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Google Fit 2024 — Monthly Step Volume & Total Distance', fontsize=15, fontweight='bold', pad=15)
fig.tight_layout()
plt.show()
```

## 🗓️ 4. Visualization 3: Activity Breakdown by Day of Week

```python
df['DayOfWeek'] = df['Date'].dt.strftime('%a')
dow_order = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
dow_avg = df.groupby('DayOfWeek')['Step_Count'].mean().reindex(dow_order)

plt.figure(figsize=(10, 5))
plt.bar(dow_avg.index, dow_avg.values, color='#4f46e5', alpha=0.9, width=0.5)
plt.axhline(10000, color='#dc2626', linestyle='--', linewidth=1.5, label='10k Goal Threshold')
plt.title('2024 Average Steps by Day of Week', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Day of Week', fontsize=11)
plt.ylabel('Average Step Count', fontsize=11)
for i, v in enumerate(dow_avg.values):
    plt.text(i, v + 200, f'{v:,.0f}', ha='center', fontweight='bold', fontsize=10)
plt.ylim(0, max(dow_avg.values) * 1.15)
plt.legend()
plt.tight_layout()
plt.show()
```
